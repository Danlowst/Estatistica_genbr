# **Fundamentos e Tipos de Dados **

Estatística Descritiva vs. Inferencial: A Descritiva tira uma "foto" do que já aconteceu (ex.: calcular a média de vendas do mês passado). A Inferencial usa uma parte dos dados para tentar "prever" o todo ou o futuro (ex.: pesquisa eleitoral).

População vs. Amostra: População é todo mundo (todos os 5.000 clientes da loja). Amostra é um pedaço representativo (200 clientes que responderam à pesquisa).


Tipos de Variáveis (Como o Python enxerga as colunas):

- Qualitativa (Categórica): É texto ou categoria (ex.: cor do produto, nível de satisfação: bom/médio/ruim).

- Quantitativa Discreta: É número inteiro que a gente conta (ex.: número de filhos, quantidade de visitas no site).

- Quantitativa Contínua: É número que a gente mede e aceita vírgula (ex.: altura, peso, temperatura, valor da venda em R$).

Média (mean)
Mediana (median)

# **Formatos de Distribuição e Correlação**

- Distribuição Normal (Curva de Sino): A maioria fica no meio, e poucos nas pontas.
Aqui, Média = Mediana = Moda (ex.: altura das pessoas, notas de uma prova equilibrada).

- Distribuição Assimétrica (Skewed): Tem uma "cauda" esticada para um lado por causa de valores extremos.

 Assimétrica à Direita (Positiva): Muitos valores baixos e poucos muito altos puxando a média para cima (Média > Mediana). Ex.: renda salarial.

Assimétrica à Esquerda (Negativa): Muitos valores altos e poucos muito baixos puxando a média para baixo (Média < Mediana). Ex.: idade de aposentadoria.

Distribuição Bimodal: Tem dois "morros" (picos). Indica que existem dois grupos diferentes misturados nos dados (ex.: tempo de maratona com atletas profissionais e amadores).

- Correlação (r): Mede se duas variáveis andam juntas (vai de -1 a +1). Positiva: tráfego no site sobe e vendas sobem.
Negativa: preço sobe e vendas caem. Lembrete importante: correlação não significa que uma coisa causou a outra!

A Média (mean)

Imagine que você e 4 amigos ganharam quantidades diferentes de balas. Um ganhou 2, outro 8, outro 10... A média é o ato de juntar todas as balas em um único monte e dividir exatamente por igual entre todo mundo.

A mediana (median)

Imagine que você colocou todas as pessoas da sala em uma fila indiana, organizadas da mais baixa para a mais alta. A mediana é exatamente a pessoa que está de pé bem no meio da fila de mãos dadas.

A moda (mode)
Sabe a "moda" das roupas, dos tênis ou dos cortes de cabelo que todo mundo está usando na escola? Na estatística, a moda é o número que mais se repete.

Amplitude (max - min)
É como esticar uma fita métrica desde o menor número até o maior número de um grupo.
2min  - 10 max = 8


Desvio Padrão (std)
Imagine uma professora (que é a média) tentando guiar um grupo de crianças no parquinho.
Se todas as crianças estão coladinhas na saia da professora, o desvio padrão é baixinho (todo mundo comportado e pertinho do centro).

Se as crianças estão correndo para os quatro cantos do parquinho (uma no escorregador, uma no balanço, uma no lanche), o desvio padrão é altíssimo (muita bagunça e espalhamento)

Ele mede o quanto os dados estão "fugindo" da média. Ajuda a saber se a situação é estável ou uma montanha-russa.

Correlação (corr)

É o "Match do Tinder" entre duas colunas de números.

Se quanto mais você estuda (Coluna A), mais sua nota aumenta (Coluna B), eles dão match perfeito: a correlação é positiva e forte (vai até +1)

Se quanto mais frio está lá fora (Coluna A), menos sorvetes você vende (Coluna B), um sobe e o outro desce: a correlação é negativa (vai até -1)

Se não tem nada a ver uma coisa com a outra (ex: número de calçado versus quantidade de chuva), a correlação é zero.

- CORRELAÇÃO POSITIVA - VÃO PARA O MESMO LADO Quando uma coisa aumenta, a outra também aumenta. Se uma coisa diminui a outra também diminui.

- CORRELAÇÃO NEGATIVA - VÃO PARA LADOS OPOSTOS Quando uma coisa aumenta, a outra diminui. Gangorra

- CORRELAÇÃO ZERO - NÃO TEM NADA A VER Quando uma coisa muda, a outra não está nem aí. Não existe nenhum padrão entre elas. Mudar um não faz diferença nenhuma no outro.

** REGRA - CORRELAÇÃO NÃO SIGNIFICA CAUSA.

Gráfico de Linhas - Evolução / tendência ao longo do tempo. Ex: Vendas mes a mes.

Barras / Colunas - Comparar valores entre categorias diferentes. Ex: Vendas do produto a vs produto b.

Pizza - Mostrar proporções (%) poucas fatias. Ex: Divisão de sócios em uma empresa.

Box plot (Caixa) - Mostrar mediana, dispersão e outliers - Ex: Comparar notas entre equipes.

Dispersão - Correlação entre duas variáveis numéricas. Ex: Investimento em anuncios x Vendas.

# **Cenário da Cafeteria Gen**

```python
# Criar base de dados simples
# Importar bibliotecas
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

dados = {
    "Dia_da_Semana":["Seg", "Ter", "Qua", "Qui", "Sex", "Sab", "Dom"], # Qualitativo
    "Cafes_Vendidos": [50,55,50,60,65,180,70], # Quantitativo Discreto, se fosse em dinheiro seria Quantitativo Continuo
    "Temperatura": [22,20,21,19,18,15,17] # Quantitativo Continuo, mas nesse exemplo em especifico, não tem virgula
}
df = pd.DataFrame(dados)
display(df)
display(df.describe())
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Temperatura',       # Eixo X: Temperatura (quanto mais à direita, mais calor)
    y='Cafes_Vendidos',    # Eixo Y: Cafés vendidos (quanto mais para cima, mais vendas)
    hue='Dia_da_Semana',   # Cor dos pontos: Muda de acordo com o dia da semana
    s=120,                 # Tamanho dos pontos para ficarem fáceis de ver
    palette='deep'         # Paleta de cores agradável
)

# 3. Adicionando títulos e melhorando a estética
plt.title('Relação entre Temperatura, Vendas de Café e Dias da Semana', fontsize=14, fontweight='bold')
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Cafés Vendidos (unidades)', fontsize=12)

# Jogando a legenda para fora do gráfico para não tampar os pontos
plt.legend(title='Dia da Semana', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adicionando linhas de grade quadriculada para facilitar a leitura
plt.grid(True, linestyle='--', alpha=0.5)

# 4. Exibindo o gráfico limpo
plt.tight_layout()
plt.show()
```

**Mini-Case 1: O Mistério do Bônus de Vendas (Média vs. Mediana)**

Cenário: Uma loja tem 5 vendedores. Em outubro, eles venderam: R$ 10.000, R$ 11.000, R$ 12.000, R$ 11.000 e R$ 56.000 (um vendedor fechou um contrato gigante único).
O gerente calculou a média (R$ 20.000) e quer exigir que todos vendam R$ 20.000 no mês seguinte.


*Como analista de dados, você concorda com essa meta? Qual medida
você mostraria ao gerente e qual é o valor dela?*

```python
dados1 = {"vendas":[10000, 11000, 12000, 11000, 56000]}
df_vendas = pd.DataFrame(dados1)

media = df_vendas["vendas"].mean()
mediana = df_vendas["vendas"].median()
amplitude = df_vendas["vendas"].max() - df_vendas["vendas"].min()

print(f"Média: {media}\n"
      f"Mediana: {mediana}\n"
      f"Amplitude: {amplitude}")

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df_vendas['vendas'],
    color='lightcoral',
    width=0.4
)

plt.title('Análise de Vendas da Equipe: O Impacto do Outlier', fontsize=12, fontweight='bold')
plt.ylabel('Valor de Vendas (R$)', fontsize=10)

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
